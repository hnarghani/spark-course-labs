```python

# ۲. لیست داده‌های ما (ورودی)
data_list = [
    "Spark is fast",
    "Spark is cool",
    "Python is great"
]

# ۳. تبدیل لیست به RDD (توزیع کردن لیست بین سرورها)
rdd = spark.sparkContext.parallelize(data_list)

# ۴. عملیات WordCount
counts = rdd.flatMap(lambda line: line.split(" ")) \
            .map(lambda word: (word, 1)) \
            .reduceByKey(lambda a, b: a + b)


```

```python
counts.collect()
```

```python
# 1. تعریف در Driver
bad_records = spark.sparkContext.accumulator(0)

def validate_data(line):
    global bad_records
    parts = line.split(",")
    
    # چک کردن اینکه آیا بخش دوم (سن) وجود دارد و عدد است یا خیر
    if len(parts) < 2 or not parts[1].isdigit():
        bad_records.add(1) # اضافه کردن به شمارنده در صورت بروز خطا
        return None
    return parts

# داده‌های شما
data = ["Ali,25", "Sara,30", "Reza,", "Mina,22", "Hamed,abc"]
rdd = spark.sparkContext.parallelize(data)

# اجرای عملیات روی سرورها
valid_rdd = rdd.map(validate_data).filter(lambda x: x is not None)
valid_rdd.collect() # این دستور باعث اجرای عملیات می‌شود

# 2. خواندن نتیجه نهایی در Driver
print(f"Total Bad Records: {bad_records.value}") 
# خروجی: Total Bad Records: 2
```
