بخش 6: توابع شرطی

```python
df.withColumn(
    "category",
    when(col("amount") >= 1000, "high")
    .otherwise("normal")
).show()
```

بخش 7: NULL Handling

```python
data2 = [
    (1, "Ali", 1200),
    (2, None, 800),
    (3, "Reza", None)
]

df2 = spark.createDataFrame(data2, ["id", "name", "amount"])
df2.show()
df2.fillna({"name": "unknown", "amount": 0}).show()
```

بخش 8: Window Function

```python
from pyspark.sql.window import Window

window_spec = Window.partitionBy("city").orderBy(col("amount").desc())

df.withColumn("row_num", row_number().over(window_spec)).show()
```

بخش 9: Join

```python
customers = spark.createDataFrame([
    (1, "Ali"),
    (2, "Sara"),
    (3, "Reza")
], ["customer_id", "name"])

orders = spark.createDataFrame([
    (1, 500),
    (1, 700),
    (2, 300)
], ["customer_id", "amount"])

customers.join(orders, on="customer_id", how="inner").show()
```
